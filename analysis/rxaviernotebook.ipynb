### EDA (Exploratory Data Analysis)

Now, for each theme park I conduct the same EDA steps, which I will explain below.

The first thing which I wanted to do was analyze wait times by season, as a lot of people don't just aim for a specific month, but want to rather visit the park in a specific season. So, I created a function 'season_convert', which takes the numerical month number and returns what season it is in. After I create a "Season" column for the dataframe, I then use a .groupby() for the day, and get the mean and median wait times. I then do the same thing for monthly and season data, grouping by month and season respectively, and still getting the mean and median wait times

```python
import numpy as np

def season_convert(month):
    if month in [12, 1, 2]: return 'Winter'
    elif month in [3, 4, 5]: return 'Spring'
    elif month in [6, 7, 8]: return 'Summer'
    else: return 'Fall'

day_order = ['Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday', 'Saturday', 'Sunday']
season_order = ['Winter', 'Spring', 'Summer', 'Fall']

# DISNEYLAND PARK
print("\nDISNEYLAND PARK")
disney_cleaned = disney_df_filtered.copy()
disney_cleaned['Date'] = pd.to_datetime(disney_cleaned['Date'])
disney_cleaned['DayOfWeek'] = disney_cleaned['Date'].dt.day_name()
disney_cleaned['Month'] = disney_cleaned['Date'].dt.month
disney_cleaned['MonthName'] = disney_cleaned['Date'].dt.strftime('%B')
disney_cleaned['Season'] = disney_cleaned['Month'].apply(season_convert)

disney_daily_stats = disney_cleaned.groupby('DayOfWeek')['Average Wait Time (mins)'].agg(['mean', 'median', 'count']).reindex(day_order)
print(disney_daily_stats.sort_values('mean'))

disney_monthly_stats = disney_cleaned.groupby(['Month', 'MonthName'])['Average Wait Time (mins)'].agg(['mean', 'median', 'count'])
disney_best_month = disney_monthly_stats['mean'].idxmin()
print(disney_monthly_stats.sort_values('mean')[['mean', 'median', 'count']])

disney_season_stats = disney_cleaned.groupby('Season')['Average Wait Time (mins)'].agg(['mean', 'median', 'count']).reindex(season_order)
print(disney_season_stats)




```

As shown above, we find that the best day for Disneyland Park is Sunday, with a mean wait time of 26.6 minutes. We also find that the best month to attend is September with a mean wait time of 22.6 minutes, and Fall has the lowest wait time out of all season with a mean wait time of 25.4 minutes.

Above, we can see that Universal seems to have the largest mean wait times and median wait times, but note that it has only 13000 observations, while the park with the second least observations, Knotts, has around 26000 observations. But, is this difference due to the fact that Universal has a much smaller data size, or actually due to the fact that Universal just has that much higher of wait times?

```python
from scipy import stats
import pandas as pd

parks = ['Disneyland', 'DCA', 'Knotts Berry Farm', 'Six Flags', 'Universal']

combined_park_data = {
    'Disneyland': disney_cleaned['Average Wait Time (mins)'].dropna(),
    'DCA': dca_cleaned['Average Wait Time (mins)'].dropna(),
    'Knotts Berry Farm': knotts_cleaned['Average Wait Time (mins)'].dropna(),
    'Six Flags': sixflags_cleaned['Average Wait Time (mins)'].dropna(),
    'Universal': universal_cleaned['Average Wait Time (mins)'].dropna()
}

welch_results = []

for park in parks:
    if park == "Universal":
        continue  # skip comparing Universal to itself
    
    t_stat, p_value = stats.ttest_ind(
        combined_park_data["Universal"],
        combined_park_data[park],
        equal_var=False
    )
    
    mean_diff = combined_park_data["Universal"].mean() - combined_park_data[park].mean()
    
    welch_results.append({
        'Comparison': f"Universal vs {park}",
        'Mean Diff (Universal - Other)': round(mean_diff, 2),
        't-statistic': round(t_stat, 4),
        'p-value': round(p_value, 6),
        'Significant (α=0.05)': 'Yes' if p_value < 0.05 else 'No'
    })

welch_df = pd.DataFrame(welch_results)
print(welch_df.to_string(index=False))
```

### Welch's t-test
Above, we ran Welch's t-test to test if Universal had longer wait times regardless of its sample size. The reason why we chose this test is because it is specifically designed to handle unequal sample sizes and unequal variances, which helps account for Universal’s smaller dataset. The results sho Universal consistently has significantly higher average wait times than every other park, with strong statistical evidence of p-values equal to zero for all comparisons of Universal to each other park.

```python
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
from scipy import stats

sns.set_style("whitegrid")
colors = ['blue', 'yellow', 'green', 'red', 'purple']
park_names = ['Disneyland', 'DCA', 'Knotts Berry Farm', 'Six Flags', 'Universal']
park_data = {
    'Disneyland': disney_cleaned['Average Wait Time (mins)'].dropna(),
    'DCA': dca_cleaned['Average Wait Time (mins)'].dropna(),
    'Knotts Berry Farm': knotts_cleaned['Average Wait Time (mins)'].dropna(),
    'Six Flags': sixflags_cleaned['Average Wait Time (mins)'].dropna(),
    'Universal': universal_cleaned['Average Wait Time (mins)'].dropna()
}

fig, axes = plt.subplots(2, 3, figsize=(18, 10))
fig.suptitle('Wait Time Distributions by Park ', fontsize=16, fontweight='bold')

for i, (park, color) in enumerate(zip(park_names, colors)):
    axis = axes[i // 3, i % 3]
    data = park_data[park]
    
    axis.hist(
        data,
        bins=50,
        color=color,
        alpha=0.6,
        edgecolor='black',
        density=True,
        label='Histogram'
    )
    
    from scipy.stats import gaussian_kde
    kde = gaussian_kde(data)
    x_range = np.linspace(data.min(), data.max(), 200)
    axis.plot(x_range, kde(x_range), color='darkred', linewidth=2.5, label='Density')
    
    axis.axvline(
        data.mean(),
        color='green',
        linestyle='--',
        linewidth=2.5,
        label=f'Mean: {data.mean():.1f}'
    )
    
    axis.axvline(
        data.median(),
        color='orange',
        linestyle='--',
        linewidth=2.5,
        label=f'Median: {data.median():.1f}'
    )
    
    axis.set_xlabel('Wait Time (mins)', fontweight='bold')
    axis.set_ylabel('Density', fontweight='bold')
    axis.set_title(f'{park}\n(n={len(data):,})', fontweight='bold')
    axis.legend(fontsize=9)
    axis.grid(True, alpha=0.3)


axes[1, 2].remove()
plt.tight_layout()
plt.show()



```

### Disneyland
Looking at Disneyland, we can see that it has a distribution that is relatively right-skewed, but has the most extreme wait times across all parks. 

### DCA (Disneyland California Adventure)
DCA has the most right-skewed distribution, with a steep drop-off. Most of the rides have short waits, but there are some outliers that pull the mean up.

### Knotts Berry Farm
Knotts has a tight distribution centered around 15-30 minutes. The mean and median are very close, and it has a small tail after 60 minutes. Overall, it has the most consistent, predictable wait times.

### Six Flags
Six Flags has the steepest drop in wait times, where a lot of rides have waits less than 10 minutes. Overall, Six Flags has a lot of rides that have short wait-times compared to the other parks, and not a lot of rides with a history of abnormally long waits.

### Universal
Universal has a bimodal distribution, and amongst all the parks, it is the widest and the most right-skewed distribution. It's peak density seems to be clustered amongst longer wait times than the other parks.
